1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
# dec_bin solo para enteros
def dec_bin_1(x):

    bin_list = [] # lista vacia donde vamos a guardar los digitos (0s o 1s) del resultado
    num = int(x) # convertimos la entrada de la función en entero (porque no sé como hacerlo con punto decimal aun)

    while num > 0: # mientras que el número sea mayor que cero hace este loop. OJO: num va a ir cambiando
        residuo = num%2 # calculamos el residuo de la primera devisión entre 2
        bin_list.append(residuo) # lo agregamos a la lista
        num = num // 2 # actualizamos num para que ahora sea el cociente de la división (sin decimales)
        
    return bin_list[::-1] # invertimos la lista
```

```python
# dec_bin para reales
def dec_bin_2(x):
    
    x = float(x)

    # Separate integer and fractional part
    integer = int(x)
    frac = x - integer

    # for the integer part
    
    if integer == 0:
        int_bits = [0]
        
    else:
        int_bits = [] # the list starts empty and we append residues like we did before
        num = integer
        while num > 0:
            residuo = num % 2
            int_bits.append(residuo)
            num = num // 2
        int_bits = int_bits[::-1]

    # for the fractional part
    
    frac_bits = [] # empty list for the fractional bits
    count = 0 
    while frac != 0 and count < 2: # we are just going to use two decimal bits
        frac = frac*2 # we multiply by two
        bit = int(frac) # we take the integer (always 0 or 1)
        frac_bits.append(bit) # we append it in our list
        frac = frac - bit # we have to repeat the process but just with the fractional part!
        count += 1 
    
    # Unimos ambas partes

    if len(frac_bits) > 0:
        int_str = ''.join(str(b) for b in int_bits)
        frac_str = ''.join(str(b) for b in frac_bits)
        return f"{int_str}.{frac_str}"

    else:
        return f"{''.join(str(b) for b in int_bits)}" # the '' before .join means join the elements with nothing in between

    #return int_bits, frac_bits
```

```python
# bin_dec solo para enteros y con máximo 8 bits
def bin_dec_1(x):

    num = str(x).zfill(8)  # 8 dígitos, rellena con ceros a la izquierda
    bin_digits = [int(digit) for digit in num]
    bin_exponents = [2**7, 2**6, 2**5,  2**4,  2**3,  2**2, 2**1, 2**0]

    prueba = list(zip(bin_digits,bin_exponents))
    final = [i*j for i,j in prueba]

    return sum(final)
```

```python
# bin_dec solo para enteros pero con n bits
def bin_dec_2(x):
    
    num = str(x)
    bin_digits = [int(digit) for digit in num]
    bin_digits = bin_digits[::-1]

    decimal = 0
    for i,bit in enumerate(bin_digits):
        decimal = decimal + bit*(2**i)
    
    return decimal
```

```python
# bin_dec para reales
def bin_dec_3(x):
    
    x = str(x)

    # Si no hay punto decimal, lo tratamos como entero
    if '.' not in x:
        integer = x[::-1]
        frac = ''
    else:
        split = x.split('.') # split me genera una lista con los elementos de x separados por el punto
        integer = split[0][::-1]
        frac = split[1]
        
   # for the integer part
    
    integer_part = 0

    for i,bit in enumerate(integer): # recuerda que enumerate produce pares i es int and bit es str
        integer_part += int(bit)*(2**i)

    # for the fractional part
        
    fractional_part = 0
    
    for i,bit in enumerate(frac, start=1): 
        fractional_part += int(bit)*(2**(-i))

    return integer_part + fractional_part
```

```python
def dec_hex(x):
    hex_list = []
    num = int(x)

    if num<=9:
        return num
    elif num==10:
        return A
    elif num==11:
        return B
    elif num==12:
        return C
    elif num==13:
        return D
    elif num==14:
        return E
    elif num==15:
        return F
    else:
        while num > 0:
            residuo = num%16
            if residuo==10:
                hex_list.append('A')
            elif residuo==11:
                hex_list.append('B')
            elif residuo==12:
                hex_list.append('C')
            elif residuo==13:
                hex_list.append('D')
            elif residuo==14:
                hex_list.append('E')
            elif residuo==15:
                hex_list.append('F')
            else:
                hex_list.append(residuo)
            num = num // 16
        
    return hex_list[::-1]
```

```python
def hex_dec(x): # pero aquí x debe ser una string 

    dec_value = 0
    potencia = 0

    for element in x[::-1]: # recorre la lista en sentido contrario
        if element == 'A':
            num = 10
        elif element == 'B':
            num = 11
        elif element == 'C':
            num = 12
        elif element == 'D':
            num = 13
        elif element == 'E':
            num = 14
        elif element == 'F':
            num = 15
        else:
            num = int(element) # es necesario volverlo entero, si no sigue pensando que es caracter?

        dec_value = dec_value + num*(16**potencia)
        
        potencia = potencia + 1 # en cada iteración aumenta una

    return dec_value
```

```python
def convert(n, b1, b2):
    '''This function converts an integer number (that has to be entered as a string) from base b1 to base b2.
    It uses decimal as intermediate representation and the possibilities for b1 or b2 are 2, 10 or 16.'''
    
    d = "0123456789ABCDEF" # possible character of the string (in order)

    # base b1 a decimal
    x = 0
    for c in n:
        x = x*b1 + d.index(c)  
        # By repeatedly factoring powers of the base, the positional representation of a number can be rewritten as
        # x_k+1 = x_k*b + d_k 

    # decimal a base b2
    if b2 == 10:
        return str(x)

    else:
        r = ""
        while x > 0:
            r = d[x % b2] + r
            x = x // b2
        return r
```

2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
# s eeeeeeee mmmmmmmmmmmmmmmmmmmmmmm

#bit 31:             s         (signo)
#bits 30–23:         eeeeeeee  (exponente, 8 bits)
#bits 22–0:          mmmmmmm...mm (mantissa, 23 bits)

def ttbits_interpretation(ttbits):
    ''' This function converts a 32 bit word into a single precision floating point,
    interpreting the various bits as sign, mantissa and exponent. Please enter 0b before the  32 bit word.'''

    # extraer el signo
    s = ttbits >> 31 # se mueve 31 a la derecha para descartar todos los bits excepto el primero
    #print(s)

    e = ttbits >> 23 # se mueve 23 a la derecha 
    e = e & 0b11111111
    #print(e)

    m = ttbits & 0b11111111111111111111111
    #print(m)

    return ((-1)**s) * (1 + m / (2**23)) * (2**(e-127))

# PENDIENTE
```

**Tips**: define two variables inizialized to 1 and halve/double them enough time to exceed the under/over-flow limits  

```python
# Overflow
x = 1.0
while x < float('inf'):
    last = x
    x *= 2
print("Overflow limit:", last)

# Underflow
y = 1.0
while y > 0.0:
    last = y
    y /= 2
print("Underflow limit:", last)
```

4\. Write a program to determine the machine precision

**Tips**: define a new variable by adding a smaller and smaller value (proceeding similarly to prob. 2) to an original variable and check the point where the two are the same 

```python
# machine precision is the the smallest number such that 1 plus that number is different to 1

x = 1.0
n = 0.1

while x+n != x:
    last = n
    n /= 2.0

print("Machine precision:", last)
```

5\. Write a function that takes in input three parameters $a$, $b$ and $c$ and prints out the two solutions to the quadratic equation $ax^2+bx+c=0$ using the standard formula:
$$
x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}
$$

(a) use the program to compute the solution for $a=0.001$, $b=1000$ and $c=0.001$

(b) re-express the standard solution formula by multiplying top and bottom by $-b\mp\sqrt{b^2-4ac}$ and again find the solution for $a=0.001$, $b=1000$ and $c=0.001$. How does it compare with what previously obtained? Why?

(c) write a function that compute the roots of a quadratic equation accurately in all cases

```python
import numpy as np
```

```python
# Solution 5a

def quadratic_solution(a,b,c):
    a,b,c = float(a),float(b),float(c)
    D = b**2 - 4*a*c
    x1 = (-b + np.sqrt(D))/(2*a)
    x2 = (-b - np.sqrt(D))/(2*a)
    return x1, x2

quadratic_solution(0.001,1000,0.001)
```

```python
# Solution 5b

def quadratic_solution_b(a, b, c):
    a,b,c = float(a),float(b),float(c)
    D = b**2 - 4*a*c

    #x1 = (-b + D)/(2*a) * (-b - D)/(-b - D)
    #x2 = (-b - D)/(2*a) * (-b + D)/(-b + D)

    x1 = 2*c/(-b-np.sqrt(D))
    x2 = 2*c/(-b+np.sqrt(D))

    return x1, x2

quadratic_solution_b(0.001,1000,0.001)
```

```python
# How does it compare with what previously obtained? Why?

# The solution in (b) is more accurate than (a).
# In (a), the result suffers from a loss of precision, whereas part (b) yields the analytically correct value.
# Why? In formula (a), we subtract two nearly equal numbers, which causes the computer to lose significant digits.
```

```python
# Solution 5c

def quadratic_stable(a, b, c):
    a, b, c = float(a), float(b), float(c)
    D = b**2 - 4*a*c
    
    # Para evitar la cancelación, siempre sumamos términos que tengan el mismo signo.
    
    if b >= 0:
        # Si b es positivo, -b y +sqrt(D) tienen signo opuesto
        # Calculamos x1 con la fórmula de (b) y x2 con la de (a)
        x1 = 2*c/(-b - np.sqrt(D))
        x2 = (-b - np.sqrt(D))/(2*a)
        
    else:
        # Si b es negativo, -b y +sqrt(D) tienen el mismo signo.
        # Calculamos x1 con la fórmula de (a) y x2 con la de (b)
        x1 = (-b + np.sqrt(D))/(2*a)
        x2 = 2*c/(-b + np.sqrt(D))
        
    return x1, x2

# Prueba con los valores del ejercicio
print(quadratic_stable(0.001, 1000, 0.001))
```

6\. Write a program that implements the function $f(x)=x(x−1)$

(a) Calculate the derivative of the function at the point $x = 1$ using the derivative definition:

$$
\frac{{\rm d}f}{{\rm d}x} = \lim_{\delta\to0} \frac{f(x+\delta)-f(x)}{\delta}
$$

with $\delta = 10^{−2}$. Calculate the true value of the same derivative analytically and compare with the answer your program gives. The two will not agree perfectly. Why not?

(b) Repeat the calculation for $\delta = 10^{−4}, 10^{−6}, 10^{−8}, 10^{−10}, 10^{−12}$ and $10^{−14}$. How does the accuracy scales with $\delta$?

```python
def func(x):
    return x*(x-1)
```

```python
# Solution 6a
# The values do not agree perfectly because the derivative definition requires delta to approach zero
# by using a finite value like $10^{-2}$, we are calculating the slope of a secant line rather than the tangent line.

delta = 10**(-2)
derivative = (func(1+delta) - func(1))/delta
derivative
```

```python
def deriv(delta,x):
    return (func(x+delta) - func(x))/delta
```

```python
# Solution 6b
# The accuracy initially improves as delta decreases because the truncation error is reduced. 
# However, for very small delta (like 10^{-14}), the accuracy decreases due to round-off error, 
# as the computer cannot handle the precision required for the difference between $f(x+\delta)$ and $f(x)$."

deltas = [10**(-2), 10**(-4), 10**(-6), 10**(-8), 10**(-10), 10**(-12), 10**(-14), 10**(-16)]
derivatives = [deriv(d,1) for d in deltas]
derivatives
```

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute? 


```python
# Solution a

import numpy as np

def semicircle(x):
    return np.sqrt(1-x**2)

N = 100
h = 2/N

x_values = np.linspace(-1 + h/2, 1 - h/2, N) # usamos h/2 para evalular la función en el punto medio del intervalo.
integral = sum(semicircle(x_values) * h)

print(f'True value: {np.pi/2}')
print(f'Riemann value (N=100): {integral}')
print(f'Error: {abs(np.pi/2 - integral)}')
```

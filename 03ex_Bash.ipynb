1\.d Find out which is the letter with most counts.

1\.e Assume an obvious numbering of the students in the file (first line is 1, second line is 2, etc.), group students "modulo 18", i.e. 1,19,37,.. 2,20,38,.. etc. and put each group in a separate file  

```python
%%bash
#a

mkdir /home/gaby/Documents/PoD/LCP/LaboratoryOfComputationalPhysics_Y8/docs_03/students
cd /home/gaby/Documents/PoD/LCP/LaboratoryOfComputationalPhysics_Y8/docs_03/students
if [ ! -f "LCP_22-23_students.csv?rlkey=47fakvatrtif3q3qw4q97p5b7" ]; then
    echo 'The file does not exist. Downloading...'
    wget "https://www.dropbox.com/scl/fi/bxv17nrbrl83vw6qrkiu9/LCP_22-23_students.csv?rlkey=47fakvatrtif3q3qw4q97p5b7"
else
    echo 'The file is already there'
fi

#b
grep 'PoD' "LCP_22-23_students.csv?rlkey=47fakvatrtif3q3qw4q97p5b7" > PoD_students.csv
grep 'Physics' "LCP_22-23_students.csv?rlkey=47fakvatrtif3q3qw4q97p5b7" > Physics_students.csv
echo 'Two new files have been created'

#c
sed '1d' "LCP_22-23_students.csv?rlkey=47fakvatrtif3q3qw4q97p5b7" | cut -d ',' -f 1 | cut -c 1 | uniq -c  > letter_count.txt
echo 'The number of students whose surname starts with each letter of the alphabet is in a document called letter_count.txt'

#d
echo 'The one with most counts is:'
sort -nr "letter_count.txt" | head -n 1

#e 
contador=0
sed '1d' "LCP_22-23_students.csv?rlkey=47fakvatrtif3q3qw4q97p5b7" | while read -r line
do
    group=$(( contador%18 ))
    echo "$line" >> "/home/gaby/Documents/PoD/LCP/LaboratoryOfComputationalPhysics_Y8/docs_03/students/group_$group.csv"
    contador=$(( contador+1 ))
done
echo "The students are now divided into 18 groups. Each in a separate file."
```
